## Q31. Armstrong numbers from 1 to 9999

For each number, I count its digits and add each digit raised to that count. If the sum equals the original number, it is an Armstrong number.

```python
armstrong_numbers = []

for number in range(1, 10000):
    digit_count = 0
    remaining = number

    while remaining > 0:
        digit_count += 1
        remaining //= 10

    digit_power_sum = 0
    remaining = number

    while remaining > 0:
        digit = remaining % 10
        digit_power_sum += digit ** digit_count
        remaining //= 10

    if digit_power_sum == number:
        armstrong_numbers.append(number)

print("Armstrong numbers:", end=" ")
print(*armstrong_numbers, sep=", ")
print("Count:", len(armstrong_numbers))
```

## Q33. Prime check and sum of primes

A number is prime if it has no divisor other than 1 and itself. To test it efficiently, I check divisors only while divisor × divisor is less than or equal to the number.

```python
n = int(input("Enter N: "))

if n < 2:
    print(n, "is not prime.")
    print("Sum of primes from 2 to N: 0")
else:
    n_is_prime = True
    divisor = 2

    while divisor * divisor <= n:
        if n % divisor == 0:
            n_is_prime = False
            break
        divisor += 1

    if n_is_prime:
        print(n, "is prime.")
    else:
        print(n, "is not prime.")

    prime_sum = 0

    for number in range(2, n + 1):
        is_prime = True
        divisor = 2

        while divisor * divisor <= number:
            if number % divisor == 0:
                is_prime = False
                break
            divisor += 1

        if is_prime:
            prime_sum += number

    print("Sum of primes from 2 to N:", prime_sum)
```

## Q34. Reverse a number without using a string

I take the last digit with `% 10` and build the reversed number with multiplication. A negative number is always reported as not a palindrome.

```python
original = int(input("Enter a number: "))
remaining = abs(original)
reversed_number = 0

while remaining > 0:
    digit = remaining % 10
    reversed_number = reversed_number * 10 + digit
    remaining //= 10

if original < 0:
    reversed_number = -reversed_number

print("Reversed number:", reversed_number)

if original >= 0 and original == reversed_number:
    print("Palindrome")
else:
    print("Not palindrome")
```

## Q35. HCF and LCM

Euclid's algorithm repeatedly replaces the two numbers with the second number and their remainder. When the second number becomes 0, the first is the HCF. LCM is `(a × b) / HCF`.

```python
a = int(input("Enter first positive integer: "))
b = int(input("Enter second positive integer: "))

if a <= 0 or b <= 0:
    print("Both numbers must be positive.")
else:
    first = a
    second = b

    while second != 0:
        first, second = second, first % second

    hcf = first
    lcm = (a * b) // hcf

    print("HCF:", hcf)
    print("LCM:", lcm)
```

## Q36. String scanner in one pass

The program visits each character only once. It counts words when a non-space character starts a new word, and classifies each character as a letter, digit, space, or special character.

```python
sentence = input("Enter a sentence: ")

characters = 0
words = 0
vowels = 0
consonants = 0
digits = 0
spaces = 0
special_characters = 0
inside_word = False

for character in sentence:
    characters += 1

    if character.isspace():
        if character == " ":
            spaces += 1
        inside_word = False
    else:
        if not inside_word:
            words += 1
            inside_word = True

        if character.lower() in "aeiou":
            vowels += 1
        elif character.isalpha():
            consonants += 1
        elif character.isdigit():
            digits += 1
        else:
            special_characters += 1

print("Total characters:", characters)
print("Total words:", words)
print("Vowels:", vowels)
print("Consonants:", consonants)
print("Digits:", digits)
print("Spaces:", spaces)
print("Special characters:", special_characters)
```

## Q38. Word Frequency Analyzer

I convert words to lowercase so “Food” and “food” count as the same word. A dictionary stores the counts, which are then displayed from most frequent to least frequent.

```python
review = input("Enter a review: ")
frequency = {}

for raw_word in review.split():
    word = raw_word.strip(".,!?;:'\"()[]").lower()

    if word:
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1

if not frequency:
    print("No words found.")
else:
    remaining_words = list(frequency.keys())
    sorted_words = []

    while remaining_words:
        most_frequent = remaining_words[0]

        for word in remaining_words:
            if frequency[word] > frequency[most_frequent]:
                most_frequent = word
            elif frequency[word] == frequency[most_frequent] and word < most_frequent:
                most_frequent = word

        sorted_words.append(most_frequent)
        remaining_words.remove(most_frequent)

    print("Word frequencies:")
    for word in sorted_words:
        print(word, ":", frequency[word])

    print("Most frequent word:", sorted_words[0])
    print("Count:", frequency[sorted_words[0]])
```

## Q41. TCS Stock Price Analysis

## Q45. Caesar Cipher Encryptor

Each English letter moves forward by the given shift. `% 26` wraps letters around the alphabet. Uppercase and lowercase are preserved, while spaces, digits, and punctuation stay unchanged.

```python
message = input("Enter a message: ")
shift = int(input("Enter shift value: "))
encrypted = ""

for character in message:
    if "A" <= character <= "Z":
        position = ord(character) - ord("A")
        new_position = (position + shift) % 26
        encrypted += chr(ord("A") + new_position)
    elif "a" <= character <= "z":
        position = ord(character) - ord("a")
        new_position = (position + shift) % 26
        encrypted += chr(ord("a") + new_position)
    else:
        encrypted += character

print("Encrypted message:", encrypted)
```

## Q46. GST Calculator Function

When the amount excludes GST, the function adds tax to the base amount. When it includes GST, the function works backward to find the base amount. It returns base, GST, and total as a tuple.

```python
def calculate_gst(amount, gst_rate=18, gst_inclusive=False):
    if gst_inclusive:
        total_amount = amount
        base_amount = amount / (1 + gst_rate / 100)
        gst_amount = total_amount - base_amount
    else:
        base_amount = amount
        gst_amount = amount * gst_rate / 100
        total_amount = base_amount + gst_amount

    return (
        round(base_amount, 2),
        round(gst_amount, 2),
        round(total_amount, 2)
    )


print(calculate_gst(1000))
print(calculate_gst(1180, gst_inclusive=True))
print(calculate_gst(500, gst_rate=5))
```

## Q47. Recursive Power Function

If the exponent is 0, the answer is 1. For a positive exponent, the function multiplies the base by the result for one smaller exponent. For a negative exponent, it returns the reciprocal.

```python
def power(base, exp):
    if exp == 0:
        return 1
    if exp < 0:
        return 1 / power(base, -exp)
    return base * power(base, exp - 1)


print("power(2, 10) =", power(2, 10))
print("power(5, 0) =", power(5, 0))
print("power(3, -2) =", power(3, -2))
```

## Q48. Identity Matrix Manipulations

A 5×5 identity matrix has 1 on its diagonal and 0 elsewhere. Multiplying by 10 changes its diagonal to 10. Its transpose looks the same because the diagonal stays in the same place.

```python
import numpy as np

identity = np.eye(5)
print("Identity matrix:")
print(identity)

multiplied = identity * 10
print("\nAfter multiplying by 10:")
print(multiplied)

# An identity matrix equals its transpose because its diagonal is unchanged.
transposed = multiplied.T
print("\nTranspose:")
print(transposed)

print("\nSum of diagonal elements:", np.trace(transposed))
print("Sum of all elements:", np.sum(transposed))
```
