## Q11. Matrix output prediction

**Predicted output:**

`[[99, 0, 0], [0, 0, 0], [0, 0, 0]]`  
`[[99, 0, 0], [99, 0, 0], [99, 0, 0]]`

**Why:** `matrix1` has three separate row lists, so only its first row changes. In `matrix2`, all three rows refer to the same list, so changing one row appears in every row.

```python
matrix1 = [[0] * 3 for _ in range(3)]
matrix2 = [[0] * 3] * 3

matrix1[0][0] = 99
matrix2[0][0] = 99

print(matrix1)
print(matrix2)
```

## Q12. `+` versus `+=` with lists

**Predicted output:**
x = [1, 2, 3, 4]
y = [1, 2, 3]
a = [1, 2, 3, 4]
b = [1, 2, 3, 4]

**Why:** `x = x + [4]` creates a new list and makes `x` refer to it; `y` still refers to the old list. `a += [4]` changes the existing list, so both `a` and `b` show the added item.

```python
x = [1, 2, 3]
y = x
x = x + [4]
print("x =", x)
print("y =", y)

a = [1, 2, 3]
b = a
a += [4]
print("a =", a)
print("b =", b)
```

## Q13. Variables after a `for` loop

**Predicted result:** `print(i)` displays `4`. `print(j)` raises a `NameError`.

**Why:** The first loop runs and leaves `i` with its last value, 4. `range(0)` has no values, so the second loop never runs and `j` is never created.

```python
for i in range(5):
    pass
print(i)

for j in range(0):
    pass
print(j)
```

## Q14. Strings and numbers

**Predicted result:**
`"10" + "20"` → `1020`
`10 + 20` → `30`
`"10" * 3` → `101010`
`10 * "3"` → `3333333333`
`"10" + 20` → `TypeError`
`int("10") + 20` → `30`

**Why:** `+` joins two strings or adds two numbers, but it cannot directly join a string and an integer. Multiplying a string by an integer repeats the string.

```python
print("10" + "20")
print(10 + 20)
print("10" * 3)
print(10 * "3")
```

```python
print("10" + 20)
```

```python
print(int("10") + 20)
```

## Q15. Short-circuit evaluation

**Predicted output:**
5
5
hello
0
10

[0]

**Why:** `or` returns its first truthy value, while `and` returns its first falsy value or the last value if all are truthy. Python stops checking as soon as the result is decided. `[0]` is truthy because the list is not empty.

```python
print(0 or 5)
print(5 or 0)
print(None or "hello")
print(0 and 5)
print(5 and 10)
print([] or "default")
print([0] or "default")
```
