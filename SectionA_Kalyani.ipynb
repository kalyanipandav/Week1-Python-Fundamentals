## Q1. Difference between `is` and `==`

`==` checks whether two objects have the same value. `is` checks whether they are the exact same object in memory.

**Example:** If `a = [1, 2]` and `b = [1, 2]`, then `a == b` is `True` because their contents match. But `a is b` is `False` because they are two separate lists.

```python
a = [1, 2]
b = [1, 2]

print(a == b)
print(a is b)
```

## Q2. Mutable and immutable types

**Mutable:** `list`, `set`, `dict`  
**Immutable:** `tuple`, `str`, `int`, `float`, `bool`, `frozenset`

Mutable means the object can be changed after it is created. For example, we can add an item to a list, but we cannot change one character of an existing string directly. Immutability helps prevent accidental changes.

## Q3. Output of `print(0.1 + 0.2 == 0.3)`

**Predicted output:** `False`

Values such as `0.1` and `0.2` cannot be stored exactly as binary floating-point numbers. Their sum has a tiny difference from `0.3`, so `==` gives `False`. This is normal floating-point behaviour, not a Python bug.

**Example:** In Python, `0.1 + 0.2` displays as `0.30000000000000004`.

```python
print(0.1 + 0.2 == 0.3)
```

## Q4. When would you choose a tuple over a list?

1. Coordinates such as (18.5, 73.8): their position values should stay together and not change accidentally.
2. A fixed RGB colour such as (255, 0, 0): it always has three ordered components.
3. A dictionary key such as (2026, 9): a tuple can be a key if its elements are immutable, but a list cannot.

## Q5. Dictionary lookup and list search

**Without “hash”:** A dictionary uses a key to reach its value directly in most cases. A list search may need to check items one by one, so the time grows with the list size.

**With “hash”:** A dictionary uses the key's hash to locate its entry, giving O(1) average lookup. Searching a list for a value takes O(n) in the worst case.

**Example:** In `{"Pune": 10, "Mumbai": 20}`, we can look up `"Pune"` by its key. In `[10, 20, 30]`, finding `30` may require checking the earlier items first.

## Q6. `range()` versus `numpy.arange()`

1. `range()` gives a range object; `numpy.arange()` gives a NumPy array.
2. `range()` uses integer steps; `numpy.arange()` can also use a decimal step.
3. A NumPy array supports element-wise numerical calculations; a range object does not support them directly.

**Example:** `list(range(3))` gives `[0, 1, 2]`. With `import numpy as np`, `np.arange(0, 2, 0.5)` gives `[0.  0.5  1.  1.5]`.

```python
import numpy as np

print(list(range(3)))
print(np.arange(3) + 2)
```

## Q7. `append()`, `extend()`, and `insert()`

Start with a fresh `a = [1, 2, 3]` and `x = [4, 5]` for each example:

- `a.append(x)` → `[1, 2, 3, [4, 5]]`. It adds `x` as one item at the end.
- `a.extend(x)` → `[1, 2, 3, 4, 5]`. It adds the items inside `x` separately.
- `a.insert(1, x)` → `[1, [4, 5], 2, 3]`. It adds `x` as one item at index 1.

## Q8. Starred unpacking

In `a, *b, c`, the first value goes to `a`, the last value goes to `c`, and the values in between become a list in `b`.

- `[10, 20, 30, 40, 50]` → `a = 10`, `b = [20, 30, 40]`, `c = 50`
- `[10, 20]` → `a = 10`, `b = []`, `c = 20`
- `[10]` → `ValueError`, because there are not enough values for both `a` and `c`

## Q9. `/`, `//`, `%`, and `**`

Using `-7` and `2`:

- `-7 / 2` → `-3.5`: division
- `-7 // 2` → `-4`: floor division, which rounds down
- `-7 % 2` → `1`: remainder
- `(-7) ** 2` → `49`: negative seven raised to the power two

**Example:** `-7 = (-4 × 2) + 1`, which helps explain why `-7 // 2` is `-4` and `-7 % 2` is `1`.


```python
print(-7 / 2)
print(-7 // 2)
print(-7 % 2)
print((-7) ** 2)
```

## Q10. Why is NumPy faster than Python lists?

NumPy keeps values of the same type in compact, continuous memory. Its calculations run in optimized compiled code, and it can work on a whole array at once instead of processing each number with a separate Python operation.

**Example:** If `arr = np.array([1, 2, 3])`, then `arr * 2` gives `[2, 4, 6]` in one array operation.
